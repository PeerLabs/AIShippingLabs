```python
import os
print("API key loaded:", bool(os.getenv("OPENAI_API_KEY")))
```

```python
from openai import OpenAI
client = OpenAI()

selected_model_name = "gpt-6-luna"
```

```python

MODEL_PRICES = {
      "gpt-6-astra": {"input": 10.00, "output": 50.00},
      "gpt-6-sol": {"input": 2.00, "output": 10.00},
      "gpt-6-luna": {"input": 0.10, "output": 0.50},
      "gpt-5.6-sol": {"input": 4.00, "output": 20.00},
      "gpt-5.6-cyber": {"input": 12.50, "output": 75.00},
      "gpt-5.3-codex": {"input": 1.75, "output": 14.00},
      "gpt-realtime-2.1": {"input": 4.00, "output": 24.00},
      "gpt-realtime-2.1-mini": {"input": 0.60, "output": 2.40},
}

def calculate_cost(model_name: str, input_tokens: int, output_tokens: int) -> float:
    prices = MODEL_PRICES[model_name.lower()]
    input_cost = (input_tokens / 1_000_000) * prices["input"]
    output_cost = (output_tokens / 1_000_000) * prices["output"]
    return input_cost + output_cost
```

```python
messages = [
    {"role": "user", "content": "tell me a bad time story about a unicorn"}
]
```

```python
response = client.responses.create(
    model=selected_model_name,
    input=messages
)
```

```python
print(response.output_text)
```

```python
print(response.usage)
```

```python
dir(response)
```

```python
usage = response.usage

cost = calculate_cost(
    model_name=selected_model_name,
    input_tokens=usage.input_tokens,
    output_tokens=usage.output_tokens
)
print(f"Cost: ${cost:.6f}")
```

```python
stream = client.responses.create(
    model=selected_model_name,
    input=messages,
    stream=True
)

response = None
for event in stream:
    if hasattr(event, 'delta'):
        print(event.delta, end='')
    if hasattr(event, 'response'):
        response = event.response
```

```python
usage = response.usage

cost = calculate_cost(
    model_name=selected_model_name,
    input_tokens=usage.input_tokens,
    output_tokens=usage.output_tokens
)
print(f"Cost: ${cost:.6f}")
```
